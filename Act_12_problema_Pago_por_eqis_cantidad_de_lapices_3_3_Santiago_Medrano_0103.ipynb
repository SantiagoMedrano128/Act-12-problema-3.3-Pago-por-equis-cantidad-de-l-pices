## Las variables
1. X --> Cantidad de lapices
2. PAG --> Pago que se realizara por los lapices

## Formulas
* $PAG = X * 0.85$
* $PAG = X * 0.85$

```python
# Para cuando es verdadero (si)
# Calcular el pago usando if else
# Dato de entrada X
# Mostrar el pago segun la decision

X = int(input("Ingrese la cantidad de lapices: "))
if X < 1000 :
 PAG = X * 0.95
else:
  PAG = X * 0.85
print("El pago a realizar es: ", PAG)
```

```python
# Para cuando es falso (NO)
# Calcular el pago usando if else
# Dato de entrada X
# Mostrar el pago segun la decision

X = int(input("Ingrese la cantidad de lapices: "))
if X < 1000 :
 PAG = X * 0.95
else:
  PAG = X * 0.85
print("El pago a realizar es: ", PAG)
print("Fin del programa")
print("Programa Realizado por Santiago Medrano NL = 37")
```
